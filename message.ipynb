```python
from langchain.agents import create_agent
from langchain_core.messages import HumanMessage
from langchain_groq import ChatGroq
import os

llm= ChatGroq(
    model="llama-3.3-70b-versatile",
    api_key=os.environ["GROQ_API_KEY"],
    timeout=120,
)

agent = create_agent(
    model=llm,
    system_prompt="You are my best friend that always jokes on me .",
)
```

```python
msg=HumanMessage(content="Hey,buddy! How are you doing today?")

result=agent.invoke({"messages": [msg]})
```

```python
print(result["messages"][-1].content)
```

```python
# message type

print(type(result["messages"][-1]))
```

```python
for message in result["messages"]:
    print(f"{message.type}: {message.content}\n")
```

# Tool Messages
Let's create a tool so agent will create some tool messages.



```python
from langchain_core.tools import tool

@tool

def fitness_coach(
    user_input: str,
) -> str:
    """
    A tool that provides fitness advice and workout routines based on user input.
    """
    # Here you can implement the logic to provide fitness advice based on the user_input.
    # For simplicity, we'll return a static response for now.
    return f"Based on your input '{user_input}', I recommend a 30-minute full-body workout routine that includes squats, push-ups, lunges, and planks."
```

```python
"""
use model that supports tools and create an agent with the fitness_coach tool. The agent will provide fitness advice and workout routines based on user input.

"""

agent1 = create_agent(
    model=llm,
    tools=[fitness_coach],
    system_prompt=" You are a certified fitness coach. You will provide fitness advice and workout routines based on user input. Use the fitness_coach tool to provide accurate and helpful responses.",
)
```

```python
result1 = agent1.invoke({"messages": "Hey my name is HK and i am 16 year old . and I want to improve my strength and endurance. Can you suggest a workout plan?"})
```
